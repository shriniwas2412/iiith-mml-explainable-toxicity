```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.ensemble import RandomForestClassifier
import numpy as np
from utils import load_data, evaluate_predictions, TARGET_COLUMNS

# 1. Load Data
train_df, val_df = load_data("train.csv")
X_train, y_train = train_df["cleaned_text"].values, train_df[TARGET_COLUMNS].values
X_val, y_val = val_df["cleaned_text"].values, val_df[TARGET_COLUMNS].values

# 2. Vectorize
print("[1/3] Extracting TF-IDF features...")
vectorizer = TfidfVectorizer(max_features=5000, stop_words="english")
X_train_vec = vectorizer.fit_transform(X_train)
X_val_vec = vectorizer.transform(X_val)

# 3. Train Model
print("[2/3] Training Random Forest Classifier...")
rf = RandomForestClassifier(n_estimators=100, max_depth=25, random_state=42, n_jobs=-1)
rf.fit(X_train_vec, y_train)

# 4. Predict & Evaluate
print("[3/3] Evaluating on validation set...")
# Predict probabilities for each label column
prob_list = rf.predict_proba(X_val_vec)
# Handle output formatting (list of arrays for each target class)
y_probs = np.column_stack([p[:, 1] if p.shape[1] > 1 else np.zeros(len(p)) for p in prob_list])

evaluate_predictions("Random Forest (TF-IDF)", y_val, y_probs)
```
