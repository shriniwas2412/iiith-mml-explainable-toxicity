```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.multiclass import OneVsRestClassifier
from utils import load_data, evaluate_predictions, TARGET_COLUMNS

# 1. Load Data
train_df, val_df = load_data("train.csv")
X_train, y_train = train_df["cleaned_text"].values, train_df[TARGET_COLUMNS].values
X_val, y_val = val_df["cleaned_text"].values, val_df[TARGET_COLUMNS].values

# 2. Vectorize
print("[1/3] Extracting TF-IDF features...")
vectorizer = TfidfVectorizer(max_features=10000, ngram_range=(1, 2), stop_words="english")
X_train_vec = vectorizer.fit_transform(X_train)
X_val_vec = vectorizer.transform(X_val)

# 3. Train Model
print("[2/3] Training Logistic Regression (One-vs-Rest)...")
clf = OneVsRestClassifier(LogisticRegression(C=4.0, max_iter=1000, random_state=42), n_jobs=-1)
clf.fit(X_train_vec, y_train)

# 4. Predict & Evaluate
print("[3/3] Evaluating on validation set...")
y_probs = clf.predict_proba(X_val_vec)
evaluate_predictions("Logistic Regression (TF-IDF)", y_val, y_probs)
```
